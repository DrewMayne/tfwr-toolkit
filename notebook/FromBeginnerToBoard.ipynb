*Whatever you came for, getting better works the same way. Two problems show it tonight: pumpkins, then sunflowers.*

### A move costs 200 ticks. A look costs 1. Thinking costs too.
*It all runs on one price list.*
- **Acting is expensive.** A move, a plant, a till or a harvest costs **200 ticks**.
- **Looking is cheap.** A question such as `can_harvest()`, `get_entity_type()` or `measure()` costs **1 tick**.
- **Thinking is not free.** Every operation your program runs, a comparison, an addition, a list lookup, costs about a tick. A clever plan is paid for in the same currency as the walking it saves.
- **So there is a balance.** A route that takes 300 ticks to work out has to save two moves just to break even. Sometimes the plain loop wins, and a clever solution is simply too heavy.
- **This is ordinary software engineering.** Know your environment, and know what each interaction with it costs: a move here; a network call, a disk read or a database query anywhere else.

*Where have you paid more for the call than for the thinking? Or the other way round?*

## Cactus: put things in order
*cactus*

Every cactus grows to a random size. `measure()` tells you the size of the cactus under the drone (a look, 1 tick), and `swap(direction)` swaps it with its neighbour that way (an action, 200 ticks). Harvest a cactus whose neighbours are in order, every cactus no smaller than the one to its west and the one to its south, and the harvest spreads: **n cacti at once pay n²** (game rule). A field in the right order pays for every cell squared; a field almost in order pays for almost nothing.

**The shape:** an order you have to reach and keep across the whole field. Every swap moves a cactus in its row and in its column at once.

## Leaderboards do not transfer

- **Each board is its own problem.** One player can be near the top of one board and far down another.
- **One drone and many drones are different problems**, even for the same crop. Alone it's routing; with a team it's coordination.
- **The best farm for a board is usually not the farm you'd use inside a full run.** A board: everything unlocked, one goal. A full run: exact amounts at the right moment, shared water and power.
- **The master board: Fastest Reset.** Start from nothing and unlock everything: every shape in one run, sharing one field, one water tank and one clock. On top of all of them it's a budgeting and ordering problem. A talk of its own.
- **Different boards reward different shapes.** Hay is throughput, cactus is sorting, the maze is search, the snake is path planning. Pick the shape you enjoy.
