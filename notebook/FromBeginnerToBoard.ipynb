## Leaderboards do not transfer

- **Each board is its own problem.** Andrew, same week: **#5** full run, **#9** cactus, **#24** maze, **#118** hay, **#209** carrots with many drones.
- **One drone and many drones are different problems**, even for the same crop. Alone it's routing; with a team it's coordination.

- **The master board: Fastest Reset.** Start from nothing and unlock everything: every shape in one run, sharing one field, one water tank and one clock. On top of all of them it's a budgeting and ordering problem. A talk of its own.

- **What transfers is the method.** The solutions don't carry over; the way of looking does.
- Neighbouring ranks are often the same result plus luck: ask whether you moved by more than the noise.
